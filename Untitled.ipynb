```python
import pandas as pd

# Change these paths as per your actual file locations
call_logs = pd.read_csv('call_logs.csv')
agent_roster = pd.read_csv('agent_roster.csv')
disposition_summary = pd.read_csv('disposition_summary.csv')

# Quick check
print("Call Logs:", call_logs.shape)
print("Agent Roster:", agent_roster.shape)
print("Disposition Summary:", disposition_summary.shape)
```

```python
# Step 2: Validation Checks

# Define the required columns
required_columns = ['call_date', 'agent_id', 'org_id']

# Check if the required columns exist in each DataFrame
for name, df in [('Call Logs', call_logs), 
                 ('Agent Roster', agent_roster), 
                 ('Disposition Summary', disposition_summary)]:
    missing_cols = [col for col in required_columns if col not in df.columns]
    if missing_cols:
        print(f"{name} is missing columns: {missing_cols}")
    else:
        print(f"{name} contains all required columns.")

# Check for missing values in the key columns
print("\nMissing values in key columns:")
print("Call Logs:")
print(call_logs[required_columns].isnull().sum())
print("Disposition Summary:")
print(disposition_summary[required_columns].isnull().sum())

# Check for duplicate rows in each dataset
print("\nDuplicate row count:")
print("Call Logs:", call_logs.duplicated().sum())
print("Agent Roster:", agent_roster.duplicated().sum())
print("Disposition Summary:", disposition_summary.duplicated().sum())
```

```python
# Step 3: Join Logic

# Merge the datasets
# Join the 'call_logs' and 'disposition_summary' on 'agent_id', 'org_id', and 'call_date'
merged_data = pd.merge(call_logs, disposition_summary, on=['agent_id', 'org_id', 'call_date'], how='left')

# Now, merge the above result with the 'agent_roster' on 'agent_id' and 'org_id' only (since 'call_date' is missing in 'agent_roster')
final_data = pd.merge(merged_data, agent_roster, on=['agent_id', 'org_id'], how='left')

# Check for any missing data in the final merged dataset
print("\nMerged Data Preview:")
print(final_data.head())

# Verify that the join was successful (check column names and a few rows)
print("\nColumns in final merged dataset:")
print(final_data.columns)
```

```python
# Step 4: Feature Engineering

# Convert 'call_date' to datetime if it's not already
final_data['call_date'] = pd.to_datetime(final_data['call_date'])

# Group by 'agent_id' and 'call_date' and calculate the required features
agent_performance = final_data.groupby(['agent_id', 'call_date']).agg(
    total_calls=('call_id', 'count'),  # Total calls made by the agent
    unique_loans_contacted=('installment_id', 'nunique'),  # Unique loans contacted
    connect_rate=('status', lambda x: (x == 'completed').mean()),  # Completed calls / Total calls
    avg_call_duration=('duration', 'mean'),  # Average call duration in minutes
    presence=('login_time', lambda x: 1 if x.notna().any() else 0)  # Presence: 1 if logged in
).reset_index()

# Check the resulting performance summary
print("\nAgent Performance Summary Preview:")
print(agent_performance.head())

# Save the report as a CSV file
agent_performance.to_csv('agent_performance_summary.csv', index=False)
```

```python
# Step 5: Create Slack-style report

# Merge with agent roster to get the agent names
agent_report = agent_performance.merge(agent_roster[['agent_id', 'users_first_name', 'users_last_name']], on='agent_id', how='left')

# Function to format Slack-style report for each agent
def create_slack_report(row):
    return (f"*Agent Name*: {row['users_first_name']} {row['users_last_name']}\n"
            f"*Total Calls Made*: {row['total_calls']}\n"
            f"*Unique Loans Contacted*: {row['unique_loans_contacted']}\n"
            f"*Connect Rate*: {row['connect_rate']*100:.2f}%\n"
            f"*Average Call Duration*: {row['avg_call_duration']:.2f} minutes\n"
            f"*Presence*: {'Yes' if row['presence'] == 1 else 'No'}\n")

# Apply the function to generate the report for all agents
agent_report['slack_report'] = agent_report.apply(create_slack_report, axis=1)

# Displaying the first few reports to verify
print("\nSlack-style Reports:")
print(agent_report[['agent_id', 'slack_report']].head())

# Optional: Save the reports to a text file
with open('slack_reports.txt', 'w') as f:
    for report in agent_report['slack_report']:
        f.write(report + "\n\n")
```
